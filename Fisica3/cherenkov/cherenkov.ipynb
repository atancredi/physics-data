```python
%matplotlib inline
 
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
 
matplotlib.rcParams['figure.dpi'] = 200

import numpy as np

```

```python
def theta_critic(n):
    return np.arcsin(1/n)

def theta_cherenkov(beta, n):
    return np.arccos(1/(beta*n))

def phi(beta, n=1.5, theta_particle = np.pi / 6):
    theta_c = theta_cherenkov(beta, n)
    return np.arccos(
        (np.cos(theta_critic(n)) - np.cos(theta_particle)*np.cos(theta_c)) / (np.sin(theta_particle) * np.sin(theta_c))
    )
```

```python
frazione_che_fugge = phi(0.9) / np.pi
frazione_totalmente_riflessa = 1 - frazione_che_fugge
frazione_totalmente_riflessa
```

```python
n_values = np.linspace(1.2,1.7, 5)
beta_values = np.linspace(0.682,1.0,100)

for n in n_values:
    frazione_totalmente_riflessa_per_beta_ed_n = [(1-(phi(beta, n)/np.pi)) for beta in beta_values]
    plt.plot(beta_values, frazione_totalmente_riflessa_per_beta_ed_n, label=f"n={n}")
plt.ylabel(r'Frazione di luce confinata')
plt.xlabel(r"Velocità della particella $\beta$")
plt.grid(True)
plt.legend()
plt.show()
```
